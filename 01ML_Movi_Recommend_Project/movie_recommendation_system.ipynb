### 3. Feature Engineering from Categorical Data

#### 3.1 Preparing Genres for Model Building
- We analyze the genres column as it is a key categorical feature for the recommendation system.
- Since it is stored in JSON-like string format, we first inspect its structure to properly extract useful genre information.


```python
import ast

def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L

movies['genres'] = movies['genres'].apply(convert)
```

```python
movies.head(1)
```

#### 3.2 Preparing keywords for Model Building

```python
movies['keywords'] = movies['keywords'].apply(convert)
movies.head(1)
```

#### 3.3 Preparing cast for Model Building

```python
movies['cast'] = movies['cast'].apply(convert)
movies.head(1)
```

```python
def convert3(obj):
    L = []
    counter = 0
    
    for i in obj:
        if counter != 3:
            L.append(i)
            counter += 1
        else:
            break         
    return L

movies['cast'] = movies['cast'].apply(convert3)
movies.head(1)
```

- Above ouput shows that only the first three cast members from each movie’s cast list.
- Limiting the cast to top actors helps reduce unnecessary data and keeps the recommendation system more focused and efficient.


#### 3.4 Preparing crew for Model Building

```python
def fetch_director(text):
    L = []
    for i in ast.literal_eval(text):
        if i['job'] == 'Director':
            L.append(i['name'])
    return L 

movies['crew'] = movies['crew'].apply(fetch_director)
```

```python

movies.head(1)  
```

#### 3.5 Removing Spaces from Multi-word Features

```python
def collapse(L):
    L1 = []
    for i in L:
        L1.append(i.replace(" ",""))
    return L1
```

```python
movies['cast'] = movies['cast'].apply(collapse)
movies['crew'] = movies['crew'].apply(collapse)
movies['genres'] = movies['genres'].apply(collapse)
movies['keywords'] = movies['keywords'].apply(collapse)
```

```python
movies.head(1)
```

#### 3.6 Transforming Overview Sentences into Word Lists

```python
movies['overview'] = movies['overview'].apply(lambda x:x.split())
```

```python
movies.head(1)
```

- This show that converts each movie overview from a sentence into a list of individual words (tokens).
- Tokenizing the text helps the recommendation system analyze and compare movie descriptions more effectively.


#### 3.7 Combining All Text Features into a Single Tags Column

```python
movies["tags"] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
movies.head(1)
```

- This show that creates a new feature called tags by combining important textual features like overview, genres, keywords, cast, and crew into a single column.
- This helps the recommendation system capture all relevant movie information in one place for better similarity matching.

#### 3.8 Final Dataset for Recommendation System

```python
new_df = movies[['movie_id', 'title', 'tags']]
new_df.head(1)
```

- This creates a new DataFrame by selecting only the essential columns: movie_id, title, and tags.
- It helps reduce unnecessary data and prepares a clean dataset focused only on information required for building the recommendation system.

```python
new_df['tags'] = new_df['tags'].apply(lambda x: " ".join(x))
new_df.head(3)
```

- This converts the tags column from a list of words into a single space-separated string.
- It is done to make the data suitable for text vectorization in the recommendation system.

#### 3.9 Converting Tags Text to Lowercase

```python
new_df['tags'] = new_df['tags'].apply(lambda x: x.lower())
```

```python
new_df.head()
```

#### 3.10 Text Normalization using Porter Stemmer

```python

import nltk
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

- This code imports the NLTK library and the PorterStemmer, which is used to reduce words to their root form (e.g., loved → love, running → run).
- It helps improve text processing in the recommendation system by treating similar words as the same token.

```python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)

new_df['tags'] = new_df['tags'].apply(stem)
```

```python
new_df.head(1)
```

### 4. Feature Extraction (Model Preparation Stage)

#### 4.1 Converting Text into Numerical Features Using CountVectorizer

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000, stop_words='english')
```

- This code initializes CountVectorizer from Scikit-learn to convert text data into numerical form by creating a bag-of-words representation.
- We limit the features to the top 5000 most frequent words and remove English stop words(is, am, are) to improve model efficiency and reduce noise.

```python
vectors = cv.fit_transform(new_df['tags']).toarray()
```

```python
vectors[0]
```

#### 4.2 Building Similarity Matrix for Recommendation System

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vectors)
```

- This code computes the cosine similarity matrix between all movie vectors.
- It measures how similar two movies are based on their text features, which helps in recommending movies with similar content.

```python
similarity.shape
```

```python
similarity[0]
```

#### 4.3 Movie Recommendation Function Using Similarity Scores

```python
def recommend(movie):
    movie_index = new_df[new_df['title'] == movie].index[0]
    distances = similarity[movie_index]
    movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:6]
    
    for i in movies_list:
        print(new_df.iloc[i[0]].title)
```

- This function recommends movies based on a given movie title by finding its index, retrieving similarity scores from the cosine similarity matrix, and sorting them in descending order.
- It then prints the top 5 most similar movies (excluding the input movie itself) from the dataset.

```python
recommend('Gandhi')
```
